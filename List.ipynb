```python
#Follows first in first out and it has two ends
```

```python
from collections import deque
```

```python
checkout = deque()
checkout.append("cusi1")
checkout.append("cusi2")
checkout.append("cusi3")
checkout.append("cusi4")
checkout.append("cusi5")
```

```python
checkout
```

```python
while checkout:
    customer = checkout.popleft()
    print("serving", customer)
```

```python
from queue import Queue
print_queue = Queue()
print_queue.put("Print the 1 pdf")
print_queue.put("Print the 2 pdf")
print_queue.put("Print the 3 pdf")
print_queue.put("Print the 4 pdf")
print_queue.put("Print the 5 pdf")



while not print_queue.empty():
    print_job = print_queue.get()
    print("printing", print_job)
```
