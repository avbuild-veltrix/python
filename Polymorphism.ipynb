Polymorphism

```python
class teacher_lecture:
    def lec_info(self):
        print("this is lec info with teacher perspective")
class student_lecture:
    def lec_info(self):
        print("This is lec info with student perspective")
```

```python
obj1 = teacher_lecture()
obj2 = student_lecture()
class_obj = [obj1,obj2]
```

```python
def par(class_obj):
    for i in class_obj:
        i.lec_info()
```

```python
par(class_obj)
```

```python
#An way to provide an option to see the private method >> A Wrapper
class Student:
    def __init__(self, name, degree):
        self.__degree = degree
        self.name = name

    def show(self):
        print("Name is", self.name, "and Degree is", self.__degree)
        
    def __Private_method(self):
        print("This is a private method")

    def acess_private_method(self):
        self.__Private_method()
```

```python
obj = Student("Kartik","Masters")
```

```python
obj.acess_private_method()
```

```python
obj.acess_private_method = "Hlo"
```

```python
obj.acess_private_method
```

```python
obj.name
```

```python
obj.show()
```

```python
#Bank Account
class Bank:
    def __init__(self, balance):
        self.__balance = balance

    def deposite(self, amount):
        self.__balance += amount

    def withdraw(self, amount):
        if self.__balance > amount:
            self.__balance -= amount
        else:
            print("Insufficient Balance")

    def showBalance(self):
        return self.__balance

acc = Bank(1000)
```
