```python
# Input values
mileage = float(input("Enter mileage of the vehicle (km/litre): "))
cost_per_litre = float(input("Enter cost of fuel per litre (Rs): "))
one_way_distance = float(input("Enter one way distance (km): "))

# Total distance (to and fro)
total_distance = 2 * one_way_distance

# Fuel required
fuel_required = total_distance / mileage

# Total fuel cost
total_cost = fuel_required * cost_per_litre

# Number of persons
persons = 6

# Amount per person
amount_per_person = total_cost / persons

# Display results
print("Amount to be paid by each person (Rs):", amount_per_person)
print(amount_per_person % 3 == 0)
```
