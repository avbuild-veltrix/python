```python
class ListOps():
    l = [1,2,3,4,5]
    def even(self, l):
        l1 = []
        for i in l:
            if i%2==0:
                l1.append(i)
        return l1
    def odd(self, l):
        l1 = []
        for i in l:
            if i%2!=0:
                l1.append(i)
        return l1
```

```python
ops1 = ListOps()
```

```python
ops1.even(ops1.l)
```
