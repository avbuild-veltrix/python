#Decorator >> Decorator is a function that modifies or extend the behaviour of the another function without changing its original code

```python
def my_decorator(func):
    def wrapper():#Wrapper is a nested function that adds the finctionality before and after calling original function.
        print("This is the line before computation")
        func()
        print("This is the line before computation")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello")
```

```python
say_hello()
```

```python
#Finding the time taken for the execution of the code
import time
def timer_decorator(func):
    def timer():
        start = time.time()
        func()
        end = time.time()
        print("The time taken for the execution of the code is", end - start)
    return timer
```

```python
@timer_decorator
def time_takes():
    print(1234564984789*987645321)
```

```python
time_takes()
```

```python
@timer_decorator
def calculation():
    print(12345 - 649 + 847 * 89 * 987 - 6453 / 21)
```

```python
calculation()
```

```python
#Class decorator
class MyDecorator:
    
    def __init__(self, func):
        self.func = func
        print("In the __init__ class")

    def __call__(self):
        print("Something is happening before func")
        self.func()
        print("Something is happening after func")
        
```

```python
@MyDecorator
def say_hello():
    print("Hello")

say_hello()
```

Some inbuilt decorator
(i). Class method >> The method that takes class itself as the first argument

```python
class Math:
    @classmethod
    def add( cls, x, y):
        return cls.__name__, x - y
```

```python
Math.add(5 , 3)
```

(ii). Static Method >> The method which can be called without creating any instance/object of the class and without using self or cls

```python
class Math:
    @staticmethod
    def add(x, y):
        return x + y
Math.add(3,5)
```

(iii). Property Decorator

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius
    @property
    def area(self):
        self.radius = radius
        return 3.14*radius**
```
