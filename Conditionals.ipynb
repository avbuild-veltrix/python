```python
x = int(input("Enter your number-x : "))
y = int(input("Enter your number-y : "))
if x > 5:
    if y > 5:
        print("Both x and y is greater than 5")
    else:
        print("x is greater than 5 but y is less than 5")
elif y > 5:
    print("x is less than 5 but y is greater than 5")
else:
    print("Both x and y is less than 5")
```
