```python
#Write a function that takes list as input and return all numeric values in a list

def only_numeric(a):
    n = []
    for i in a:
        if type(i) == int or type(i) == float:
            n.append(i)
    return n
```

```python
 only_numeric([1,2,3,4,"avi","kartik","indar"])
```

```python
#Now in input take nested list as well
only_numeric([1,2,3.3,True, "pwskills", "Kartik", 5+8j, ["Najar",4,8]])
```

```python
def only_numeric(a):
    n = []
    for i in a:
        if type(i) == list:
            for j in i:
                if type(j) == int or type(j) == float:
                    n.append(j)
        else:
            if type(i) == int or type(i) == float:
                n.append(i)
    return n
```

```python
only_numeric([1,2,3.3,True, "pwskills", "Kartik", 5+8j, ["Najar",4,8]])
```
