```python
class Student:
    def __init__(self, name):
        self.name = name # self means this variable/methid is referring to current instance of the class
    @classmethod #bound to the class
    def student_name(cls, name):# instead of first parameter you are using cls>> meaning this metod is associated with class itself
        return cls(name)
```

```python
obj1 = Student.student_name("Kartik")
```

```python
obj1.name
```

```python
#Class method can modify class state applicable to all the instances of the class
class Student:
    total_student = 0#Class variable
    def __init__(self, name):#Instance Method
        self.name = name#Instance variable
        Student.total_student = Student.total_student + 1 #total_student  is the class variable associated with all the instances
    @classmethod
    def get_total_student(cls):
        return cls.total_student
```

```python
Student.total_student
```

```python
obj1 = Student("Kartik")
obj2 = Student("Avinash")
```

```python
obj1.name
```

```python
obj2.name
```
